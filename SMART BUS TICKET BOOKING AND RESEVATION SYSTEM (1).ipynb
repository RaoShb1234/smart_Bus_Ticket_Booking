```python
# Smart Bus Reservation System

cities = ["Karachi", "Hyderabad", "Sukkur", "Mirpur Khas", "Lahore"]

routes = [
    {"id": "R101", "from": "Karachi", "to": "Hyderabad", "time": "08:00 AM", "price": 1200, "seats": 40},
    {"id": "R102", "from": "Karachi", "to": "Sukkur", "time": "09:30 AM", "price": 2000, "seats": 36},
    {"id": "R103", "from": "Hyderabad", "to": "Mirpur Khas", "time": "11:00 AM", "price": 800, "seats": 32},
    {"id": "R104", "from": "Karachi", "to": "Lahore", "time": "07:00 PM", "price": 5000, "seats": 45},
    {"id": "R105", "from": "Sukkur", "to": "Hyderabad", "time": "02:00 PM", "price": 1500, "seats": 30}
]

bookings = []


def find_route(route_id):
    for route in routes:
        if route["id"] == route_id:
            return route
    return None


def show_routes():
    print("\nBUS ROUTES")
    for route in routes:
        print(route["id"], "|", route["from"], "to", route["to"],
              "| Time:", route["time"], "| Fare: Rs.", route["price"],
              "| Seats left:", route["seats"])


def search_route():
    search = input("Enter route ID or city name: ").strip().lower()
    found = False

    for route in routes:
        if (route["id"].lower() == search or
                route["from"].lower() == search or
                route["to"].lower() == search):
            print(route["id"], "|", route["from"], "to", route["to"],
                  "| Time:", route["time"], "| Fare: Rs.", route["price"],
                  "| Seats left:", route["seats"])
            found = True

    if found == False:
        print("No route found.")


def book_ticket():
    print("\nBOOK A TICKET")
    name = input("Enter passenger name: ").strip()

    if name == "":
        print("Please enter a name.")
        return

    age_text = input("Enter passenger age: ").strip()
    if not age_text.isdigit() or int(age_text) <= 0:
        print("Please enter a valid age.")
        return
    age = int(age_text)

    route_id = input("Enter route ID (for example R101): ").strip().upper()
    route = find_route(route_id)

    if route == None:
        print("Route not found.")
        return

    seats_text = input("How many seats do you want? ").strip()
    if not seats_text.isdigit() or int(seats_text) <= 0:
        print("Please enter a valid number of seats.")
        return
    seats = int(seats_text)

    if seats > route["seats"]:
        print("Sorry, only", route["seats"], "seats are left.")
        return

    booking_id = "B" + str(1001 + len(bookings))
    total = route["price"] * seats

    booking = {
        "id": booking_id,
        "name": name,
        "age": age,
        "route": route_id,
        "seats": seats,
        "total": total,
        "status": "Confirmed"
    }

    bookings.append(booking)
    route["seats"] = route["seats"] - seats

    print("\nTicket booked successfully!")
    print("Booking ID:", booking_id)
    print("Passenger:", name)
    print("Route:", route["from"], "to", route["to"])
    print("Seats:", seats)
    print("Total fare: Rs.", total)


def show_bookings():
    print("\nMY BOOKINGS")

    if len(bookings) == 0:
        print("No bookings yet.")
        return

    for booking in bookings:
        print(booking["id"], "|", booking["name"],
              "| Route:", booking["route"], "| Seats:", booking["seats"],
              "| Fare: Rs.", booking["total"], "|", booking["status"])


def cancel_booking():
    booking_id = input("Enter booking ID: ").strip().upper()

    for booking in bookings:
        if booking["id"] == booking_id:
            if booking["status"] == "Cancelled":
                print("This booking is already cancelled.")
                return

            route = find_route(booking["route"])
            route["seats"] = route["seats"] + booking["seats"]
            booking["status"] = "Cancelled"
            print("Booking cancelled.")
            print("Your seats are available again.")
            return

    print("Booking ID not found.")


def show_seats():
    print("\nSEAT AVAILABILITY")
    for route in routes:
        if route["seats"] == 0:
            status = "Fully booked"
        else:
            status = "Seats available"

        print(route["id"], "|", route["from"], "to", route["to"],
              "|", route["seats"], "seats left |", status)


def booking_report():
    confirmed = 0
    cancelled = 0
    money = 0
    seats_booked = 0

    for booking in bookings:
        if booking["status"] == "Confirmed":
            confirmed = confirmed + 1
            money = money + booking["total"]
            seats_booked = seats_booked + booking["seats"]
        else:
            cancelled = cancelled + 1

    print("\nBOOKING REPORT")
    print("Total bookings:", len(bookings))
    print("Confirmed bookings:", confirmed)
    print("Cancelled bookings:", cancelled)
    print("Seats booked right now:", seats_booked)
    print("Total confirmed fare: Rs.", money)


while True:
    print("\nSMART BUS RESERVATION SYSTEM")
    print("1. Show bus routes")
    print("2. Search a route")
    print("3. Book a ticket")
    print("4. Show bookings")
    print("5. Cancel a booking")
    print("6. Check seats")
    print("7. Booking report")
    print("8. Exit")

    choice = input("Choose an option (1-8): ").strip()

    if choice == "1":
        show_routes()
    elif choice == "2":
        search_route()
    elif choice == "3":
        book_ticket()
    elif choice == "4":
        show_bookings()
    elif choice == "5":
        cancel_booking()
    elif choice == "6":
        show_seats()
    elif choice == "7":
        booking_report()
    elif choice == "8":
        print("Thanks for using the bus reservation system!")
        break
    else:
        print("Invalid choice. Please enter a number from 1 to 8.")
```
